# Demonstração de inferência
Este notebook mostra como carregar o vetorizador e o modelo salvos e fazer previsões em textos novos.

```python
from pathlib import Path
import joblib
import pandas as pd

model_path = Path('modelos/modelo_tfidf_logistica.joblib')
vet_path = Path('modelos/vetorizador_tfidf.joblib')

vet = joblib.load(vet_path)
model = joblib.load(model_path)

texts = [
    'Exemplo de notícia: governo anuncia novo programa.',
    'Confira os boatos que circulam nas redes sociais.'
]
df = pd.DataFrame({'texto': texts})
X = vet.transform(df.texto)
probs = model.predict_proba(X)[:, list(model.classes_).index(1)]
preds = (probs >= 0.5).astype(int)
df['probabilidade_verdadeiro'] = probs
df['rotulo_previsto'] = preds
df
```
